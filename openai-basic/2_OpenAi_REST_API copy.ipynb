# Responses API Endpoint

```python
import requests

headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

payload = {
    "model": "gpt-5-nano",
    "input": [
        {
            "role": "user",
            "content": "재미있는 사실 하나 알려주세요"
        }
    ]
}

payload
```

## 반환값은 JSON

```python
response = requests.post(
    "https://api.openai.com/v1/responses",
    headers=headers,
    json=payload
)

response.json()
```

```python
response.json()["output"][1]["content"][0]["text"]
```

# chat completions API Endpoint

```python
import requests

headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

payload = {
    "model": "gpt-5-nano",
    "messages": [
        {"role": "user",   "content": "재미있는 사실 하나 알려주세요"}]
}
```

```python
response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,
    json=payload
)

response.json()
```

```python
response.json()["choices"][0]["message"]["content"]
```

# What is the openai package?

It's known as a Python Client Library.

It's nothing more than a wrapper around making this exact call to the http endpoint.

It just allows you to work with nice Python code instead of messing around with janky json objects.

But that's it. It's open-source and lightweight. Some people think it contains OpenAI model code - it doesn't!


# Responses API

```python
# Create OpenAI client
from openai import OpenAI

openai = OpenAI()

response = openai.responses.create(
    model="gpt-5-nano",
    input= "재미있는 사실 하나 알려주세요"
)

print(response.output[1].content[0].text)

```

# openai.chat.completions API

```python
# Create OpenAI client

from openai import OpenAI
openai = OpenAI()

response = openai.chat.completions.create(model="gpt-5-nano", messages=[{"role": "user", "content": "재미있는 사실 하나 알려주세요"}])

response.choices[0].message.content
```

# OpenAI Compatible Endpoints

```python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

load_dotenv(override=True)

google_api_key = os.getenv("GOOGLE_API_KEY")

if not google_api_key:
    print("No API key was found - please be sure to add your key to the .env file, and save the file! Or you can skip the next 2 cells if you don't want to use Gemini")
elif not google_api_key.startswith("AIz"):
    print("An API key was found, but it doesn't start AIz")
else:
    print("API key found and looks good so far!")

```

```python
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)

response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=[{"role": "user", "content": "재미있는 사실 하나 알려주세요"}])

response.choices[0].message.content
```
