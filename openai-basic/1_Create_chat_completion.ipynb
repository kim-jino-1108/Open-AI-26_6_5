# Connecting to OpenAI 

````python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
````



The next cell is where we load in the environment variables in your `.env` file and connect to OpenAI.  



````python
# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

# Check the key

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
````

### Let's make a quick call to a Frontier model 

````python
# To give you a preview -- calling OpenAI with these messages is this easy. Any problems, head over to the Troubleshooting notebook.

message = "안녕하세요, GPT! 처음으로 메시지를 보내드립니다! 반갑습니다!"

messages = [{"role": "user", "content": message}]

messages
````

````python
openai = OpenAI()

response = openai.chat.completions.create(model="gpt-5-nano", messages=messages)
response.choices[0].message.content
````

````python
print(response)
````

## Types of prompts



**A system prompt** that tells them what task they are performing and what tone they should use

**A user prompt** -- the conversation starter that they should reply to

Messages

The API from OpenAI expects to receive messages in a particular structure.
Many of the other APIs share this structure:

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```


````python
messages = [
    {"role": "system", "content": "당신은 친절한 어시스턴트입니다"},
    {"role": "user", "content": "2 + 2는 무엇인가요?"}
]

response = openai.chat.completions.create(model="gpt-4.1-nano", messages=messages)
response.choices[0].message.content
````

# 1. system role - 일반적인 어시스턴트 행동 지침

````python
# 1. system role - 일반적인 어시스턴트 행동 지침
messages_system = [
    {"role": "system", "content": "당신은 친절하고 공손한 어시스턴트입니다. 항상 존댓말을 사용하세요."},
    {"role": "user", "content": "파이썬에서 리스트를 정렬하는 방법을 알려줘"}
]

response_system = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages_system
)
print("=== system role 응답 ===")
print(response_system.choices[0].message.content)
````

# 2. developer role - 개발자 수준의 고급 지침 (system보다 높은 우선순위)

````python
# 2. developer role - 개발자 수준의 고급 지침 (system보다 높은 우선순위)
messages_developer = [
    {"role": "developer", "content": "너는 시니어 백엔드 개발자야. 코드 예제는 반드시 타입 힌트와 docstring을 포함해야 해. 설명은 최소화하고 코드 위주로 답변해."},
    {"role": "user", "content": "파이썬에서 리스트를 정렬하는 방법을 알려줘"}
]

response_developer = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages_developer
)
print("\n=== developer role 응답 ===")
print(response_developer.choices[0].message.content)
````

# 예제 1: 응답 제한 (developer가 system을 덮어씀)

````python
#예제 1: 응답 제한 (developer가 system을 덮어씀)
# system은 "무엇이든 답해"라고 하지만
# developer는 "음식만 답해"라고 제한 → developer 우선
messages = [
    {"role": "system", "content": "당신은 무엇이든 친절하게 답변하는 어시스턴트입니다."},
    {"role": "developer", "content": "오직 음식 관련 질문에만 답변하라. 다른 주제는 거절해라."},
    {"role": "user", "content": "오늘 날씨 어때?"}
]
response_developer = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)
print("\n=== developer role 응답 ===")
print(response_developer.choices[0].message.content)
````

# 예제 2: 언어 강제 설정

````python
# system은 한국어로 답하라고 하지만
# developer는 무조건 영어로만 답하라고 지시
messages = [
    {"role": "system", "content": "항상 한국어로 답변하세요."},
    {"role": "developer", "content": "Respond in English only. Never use Korean under any circumstances."},
    {"role": "user", "content": "안녕하세요! 오늘 기분이 어때요?"}
]
response_developer = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)
print("\n=== developer role 응답 ===")
print(response_developer.choices[0].message.content)
````

# 예제 3: 페르소나 충돌

````python
# system은 "친절한 상담사"
# developer는 "냉정하고 직설적으로"
messages = [
    {"role": "system", "content": "당신은 따뜻하고 공감 능력이 뛰어난 심리 상담사입니다. 항상 위로의 말을 건네세요."},
    {"role": "developer", "content": "감정적인 표현 없이 팩트만 냉정하게 전달하라. 위로나 공감 표현 금지."},
    {"role": "user", "content": "요즘 너무 힘들고 지쳐요..."}
]
response_developer = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)
print("\n=== developer role 응답 ===")
print(response_developer.choices[0].message.content)
````
