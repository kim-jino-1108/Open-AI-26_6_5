# openai.chat.completions Paramerer

## 1.  temperature

```python
from openai import OpenAI
openai = OpenAI()

# temperature 비교 테스트
prompt = "내일 날씨가 어떨지 시적으로 묘사해주세요"  # 창의성 차이가 잘 드러나는 주제

temperatures = [0.0, 0.7, 1.5]
labels = ["🧊 차갑고 딱딱하게 (0.0)", "⚖️ 균형잡힌 (0.7)", "🔥 매우 창의적 (1.5)"]

for temp, label in zip(temperatures, labels):
    response = openai.chat.completions.create(
        model="gpt-4.1-nano",
        messages=[{"role": "user", "content": prompt}],
        temperature=temp
    )
    print(f"\n=== {label} ===")
    print(response.choices[0].message.content)
    print("-" * 50)
```

## 2.max_completion_tokens

```python
from openai import OpenAI
openai = OpenAI()

# max_completion_tokens 비교 테스트
prompt = "인공지능의 역사에 대해 설명해주세요"  # 길게 답할 수 있는 주제

token_limits = [20, 100, 500]
labels = ["🔴 매우 짧게 (20)", "🟡 중간 길이 (100)", "🟢 충분한 길이 (500)"]

for max_tokens, label in zip(token_limits, labels):
    response = openai.chat.completions.create(
        model="gpt-4.1-nano",
        messages=[{"role": "user", "content": prompt}],
        max_completion_tokens=max_tokens
    )
    result = response.choices[0].message.content
    finish_reason = response.choices[0].finish_reason  # 종료 이유 확인

    print(f"\n=== {label} ===")
    print(f"응답: {result}")
    print(f"종료 이유: {finish_reason}")  # 'stop' = 정상종료 / 'length' = 토큰 초과로 잘림
    print("-" * 50)
```

## 3. top_p

```python
from openai import OpenAI
openai = OpenAI()

# top_p 비교 테스트
prompt = "용이 사는 마법의 숲을 묘사해주세요"  # 창의적 표현 차이가 잘 드러나는 주제

top_p_values = [0.1, 0.5, 1.0]
labels = ["🎯 보수적 (0.1) - 확률 높은 단어만", "⚖️ 균형 (0.5) - 적당한 다양성", "🌈 다양성 최대 (1.0) - 모든 단어 가능"]

for top_p, label in zip(top_p_values, labels):
    response = openai.chat.completions.create(
        model="gpt-4.1-nano",
        messages=[{"role": "user", "content": prompt}],
        top_p=top_p,
        temperature=1.0  # top_p 효과만 보기 위해 temperature 고정
    )
    print(f"\n=== {label} ===")
    print(response.choices[0].message.content)
    print("-" * 50)
```

## 4.presence_penalty

```python
from openai import OpenAI
openai = OpenAI()

# presence_penalty 비교 테스트
prompt = "봄, 여름, 가을, 겨울 각 계절을 한 단어로 표현하고 이유를 설명해주세요"  # 주제 다양성 차이가 잘 드러나는 주제

penalty_values = [-1.0, 0.0, 1.5]
labels = [
    "🔁 반복 허용 (-1.0) - 같은 주제 계속 언급",
    "⚖️ 기본값 (0.0)  - 패널티 없음",
    "🚀 새 주제 강제 (1.5) - 다양한 표현 유도"
]

for penalty, label in zip(penalty_values, labels):
    response = openai.chat.completions.create(
        model="gpt-4.1-nano",
        messages=[{"role": "user", "content": prompt}],
        presence_penalty=penalty,
        temperature=0.7  # presence_penalty 효과만 보기 위해 temperature 고정
    )
    print(f"\n=== {label} ===")
    print(response.choices[0].message.content)
    print("-" * 50)
```

## 5. frequency_penalty

```python
from openai import OpenAI
openai = OpenAI()

# frequency_penalty 비교 테스트
prompt = "사과에 대해 3문장으로 설명해주세요"  # 짧고 단어 반복이 잘 보이는 주제

penalty_values = [0.0, 2.0]
labels = [
    "⚖️ 기본값 (0.0)",
    "✨ 반복 억제 (2.0)"
]

for penalty, label in zip(penalty_values, labels):
    response = openai.chat.completions.create(
        model="gpt-4.1-nano",
        messages=[{"role": "user", "content": prompt}],
        frequency_penalty=penalty,
        max_completion_tokens=80  # 응답 길이 제한으로 실행시간 단축
    )
    print(f"\n=== {label} ===")
    print(response.choices[0].message.content)
    print("-" * 40)
```

# Responses API Parameter

## reasoning

```python
from openai import OpenAI
client = OpenAI()

response = client.responses.create(
    model="o3-mini",
    input="다람쥐는 도토리를 얼마나 많이 모을 수 있을까요?",
    reasoning={
        "effort": "high"
    }
)

print(response)
print(response.output[1].content[0].text)
```
