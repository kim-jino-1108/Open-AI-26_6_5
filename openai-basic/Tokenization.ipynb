```python
import nltk
nltk.download('punkt')

from nltk.tokenize import word_tokenize, sent_tokenize

text = "Dr. Smith went to the U.S.A. He loves NLP!"

# 단어 토크나이징
word_tokenize(text)
# ['Dr.', 'Smith', 'went', 'to', 'the', 'U.S.A.', 'He', 'loves', 'NLP', '!']

# 문장 토크나이징
sent_tokenize(text)
# ['Dr. Smith went to the U.S.A.', 'He loves NLP!']
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

text = "Apple is looking at buying U.K. startup for $1 billion."
doc = nlp(text)

# 토큰
[token.text for token in doc]
# ['Apple', 'is', 'looking', 'at', 'buying', 'U.K.', 'startup', 'for', '$', '1', 'billion', '.']

# 품사 태깅도 동시에
[(token.text, token.pos_) for token in doc]
# [('Apple', 'PROPN'), ('is', 'AUX'), ('buying', 'VERB'), ...]

# 개체명 인식
[(ent.text, ent.label_) for ent in doc.ents]
# [('Apple', 'ORG'), ('U.K.', 'GPE'), ('$1 billion', 'MONEY')]
```

```python
from transformers import BertTokenizer

tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

text = "I am playing football"
tokens = tokenizer.tokenize(text)
# ['i', 'am', 'play', '##ing', 'foot', '##ball']

# ID로 변환
ids = tokenizer.encode(text)
# [101, 1045, 2572, 2652, 2374, 102]
#  ^CLS                           ^SEP

# 전체 인코딩 (모델 입력용)
encoding = tokenizer.encode_plus(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True,
    max_length=128
)
```

```python
from tokenizers import ByteLevelBPETokenizer

# 학습
tokenizer = ByteLevelBPETokenizer()
tokenizer.train(files=["corpus.txt"], vocab_size=5000, min_frequency=2)

# 사용
output = tokenizer.encode("I am playing football")
print(output.tokens)
# ['I', 'Ġam', 'Ġplay', 'ing', 'Ġfoot', 'ball']
# "Ġ" = 공백 포함 의미 (GPT 방식)
```

```python
import sentencepiece as spm

# 학습
spm.SentencePieceTrainer.train(
    input="corpus.txt",
    model_prefix="my_model",
    vocab_size=8000,
    model_type="bpe"       # 또는 "unigram", "word", "char"
)

# 사용
sp = spm.SentencePieceProcessor()
sp.load("my_model.model")

sp.encode("I love NLP", out_type=str)
# ['▁I', '▁love', '▁NL', 'P']
# "▁" = 단어 시작 공백 표시

sp.encode("I love NLP", out_type=int)
# [12, 456, 1023, 89]
```
